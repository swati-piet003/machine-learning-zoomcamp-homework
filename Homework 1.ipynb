```python
#Find the maximum fuel efficiency for cars from Asia
asia_max_fuel_efficiency = df[df["origin"] == "Asia"]["fuel_efficiency_mpg"].max()
print(asia_max_fuel_efficiency)
```

```python
median_before = df["horsepower"].median()
print("Median horsepower before filling missing values:", median_before)
```

```python
#most frequent value of horsepower
most_frequent_horsepower = df["horsepower"].mode()[0]
print("Most frequent horsepower:", most_frequent_horsepower)
```

```python
df["horsepower"] = df["horsepower"].fillna(most_frequent_horsepower)
```

```python
median_after = df["horsepower"].median()
print("Median horsepower after filling missing values:", median_after)
```

```python
asia_cars = df[df["origin"] == "Asia"]
print(asia_cars.shape[0])
selected_columns = asia_cars[["vehicle_weight", "model_year"]]
first_seven = selected_columns.head(7)
print(first_seven)
```

```python
#Convert the table into a NumPy array called X
X = first_seven.to_numpy()
print(X)
```

```python
XTX = X.T @ X
print(XTX)
XTX_inverse = np.linalg.inv(XTX)
print(XTX_inverse)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inverse @ X.T @ y
print(w)
print(w.sum())
print("Regression weights:", w)
print("Sum of weights:", w.sum())
```
